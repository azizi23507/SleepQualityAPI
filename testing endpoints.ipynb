```python
import requests
```

```python
body = {
  "name": "Ben Fischer",
    "age": 41,
    "email": "ben.fischer@example.com"
}
response = requests.post("http://localhost:8000/patients", json=body)
print(response.json())
print(response.status_code)
```

```python
response = requests.get("http://localhost:8000/patients/1")
print(response.json())
print(response.status_code)
```
